```python
import matplotlib.pyplot as plt
y1=[]
y2=[]
x=range(-100,100,10)
for i in x:y1.append(i**2)
for i in x:y2.append(-i**2)
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000)
plt.axhline(0)
plt.axvline(0)
plt.savefig("quad.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print(df.head())
```

```python
plt.subplot(2, 3, 1)
plt.plot(df['sepal length (cm)'], df['petal length (cm)'])
plt.subplot(2, 3, 2)
plt.plot(df['sepal width (cm)'], df['petal width (cm)'])
plt.subplot(2, 3, 3)
plt.scatter(df['sepal length (cm)'], df['petal length (cm)'], c='green')
plt.title('subplot')
plt.show()
```

```python
ages=[2,5,70,40,30,45,50,45,43,40,44,60,7,13,57,18,90,77]
range=(0,100)
bins=10
plt.hist(ages,bins,range,color='cyan',histtype='bar',rwidth=0.8,alpha=0.5)
plt.xlabel('age')
plt.ylabel('No.of people')
plt.title("My histogram")
plt.show()
```
